## 13. Outlier Detection

Check for unusually high or low values in selected numerical columns. This will help determine whether any observations need further investigation before analysis and model development.

```python
check_cols = [
    "Age",
    "DistanceFromHome",
    "TotalWorkExperienceInYears",
    "ExperienceYearsAtThisCompany",
    "EmpLastSalaryHikePercent"
]

for col in check_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outlier_count = ((df[col] < lower_bound) | (df[col] > upper_bound)).sum()

    print(f"{col}: {outlier_count} potential outliers")
```

### Finding
The IQR method identified 51 potential outliers in `TotalWorkExperienceInYears` and 56 in `ExperienceYearsAtThisCompany`. No potential outliers were detected in the other three columns. These values are retained because they represent long-serving employees, are consistent with employee age (the logical checks found no impossible values), and are not data errors. Removing them would also discard real employees from a dataset of only 1,200 records.

## 14. Data Cleaning Decision

After checking the data, I found nothing that needed fixing. There were no missing values, no duplicate employees, no inconsistent labels, and no impossible values. The few unusually high experience values belong to long-serving employees, so I kept them. Because of this, the dataset is saved as it is, and the original raw file in `data/raw` stays untouched.

## 15. Feature Engineering

In this step, a new feature named `ExperienceBeforeCurrentCompany` is created by subtracting the employee's experience at the current company from their total work experience.

```python
df["ExperienceBeforeCurrentCompany"] = (
    df["TotalWorkExperienceInYears"]
    - df["ExperienceYearsAtThisCompany"]
)
```
