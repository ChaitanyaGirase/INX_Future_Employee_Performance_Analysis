# Employee Performance Prediction

## Objective

The objective of this notebook is to predict employee performance ratings using the trained Gradient Boosting model.

The model was trained and evaluated in `train_model.ipynb`. In this notebook, we will load the saved model, prepare employee data, and predict performance ratings based on the available employee information.

## 1. Import Libraries

The required libraries are imported to load the saved model, prepare employee data, and display prediction results.

```python
import pandas as pd 
import joblib 
from pathlib import Path
```

## 2. Load the Trained Model

In this step, we load the Gradient Boosting model that was trained and saved in `train_model.ipynb`.

Loading the saved pipeline allows us to make predictions without training the model again. It also ensures that the same preprocessing steps used during training are applied to the input data.

```python
model = joblib.load("../../src/models/gradient_boosting_final.pkl")
```

## 3. Prepare Employee Data for Prediction

In this step, we load the processed employee dataset and prepare the input features required by the trained model.

The target column, `PerformanceRating`, is kept separate because the purpose is to predict it. Employee identifiers and `Attrition` are excluded because they were not used as input features during model training.

The input data must contain the same features used when training the model.

```python
df = pd.read_csv("../../data/processed/employee_performance_processed.csv")

print("Dataset shape:", df.shape)
```

## 4. Separate Input Features and Target

In this step, we separate the employee information used for prediction from the actual performance ratings.

The `PerformanceRating` column is the target we want to predict, so it must not be included in the input features. We also exclude `EmpNumber` and `Attrition` because they were not used during model training.

The remaining columns will be used as input features for the trained model.

```python
y_actual = df["PerformanceRating"]

X = df.drop(columns=["PerformanceRating", "EmpNumber", "Attrition"])
```

```python
print("Input features shape:", X.shape)
print("Actual target shape:", y_actual.shape)

print("\n Input features:\n", X.columns.to_list())
```

## 5. Predict Employee Performance Ratings

In this step, we use the saved Gradient Boosting model to predict employee performance ratings based on the prepared input features.

The model applies the same preprocessing steps used during training and generates a predicted rating for each employee. We will then compare these predictions with the actual ratings to understand how the model performs on the processed dataset.

```python
y_pred = model.predict(X)
```

```python
print("Number of predictions:", len(y_pred))

print("\nFirst 20 predicted ratings:", y_pred[:20])
```

## 6. Compare Actual and Predicted Ratings

In this step, we create a table containing the actual performance ratings and the ratings predicted by the trained model.

This helps us inspect individual predictions and understand where the model's predictions match or differ from the actual ratings.

```python
prediction_results = pd.DataFrame({
    "Actual Performance Rating": y_actual, 
    "Predicted Performance Rating": y_pred
})

prediction_results.head(20)
```

## 7. Check Prediction Results

In this step, we identify which predicted performance ratings match the actual ratings and which do not.

This helps us inspect the model's predictions across the dataset.

```python
prediction_results["Prediction Correct"] = prediction_results["Actual Performance Rating"] == prediction_results["Predicted Performance Rating"]
```

```python
print("Prediction Results:", prediction_results["Prediction Correct"].value_counts())

print("\n First 10 prediction results:\n", prediction_results.head(10))
```

## 8. Conclusion

In this notebook, we loaded the trained Gradient Boosting model and used it to predict employee performance ratings from the processed dataset.

The model generated predictions for all 1,200 employee records. We compared the predicted ratings with the actual ratings to inspect the results.

The comparison showed 1,183 correct predictions and 17 incorrect predictions. However, these results are based on the same dataset used during model development and should not be treated as an independent measure of model performance.

During testing, the model achieved an accuracy of 92.92% on the held-out test dataset. This result provides a more reliable estimate of how the model performs on unseen data.

The model demonstrates the potential of machine learning for analysing employee performance patterns. Its predictions should be used as a supporting analytical tool rather than as the sole basis for employee-related decisions.
